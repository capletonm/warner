```python
df = pd.read_csv('metaclean_cleaned.csv')

# Bij de 2 films met meerdere regisseurs pakken we de eerste
df["director_first"] = df["director"].str.split(",").str[0].str.strip()

# DIRECTOR-tabel: unieke regisseurs met een eigen ID
unique_directors = sorted(df["director_first"].unique())
director = pd.DataFrame({
    "director_id": range(1, len(unique_directors) + 1),
    "director_name": unique_directors,
})

# director_id als FK in MOVIE; de naam hoort alleen in DIRECTOR
df = df.merge(director, left_on="director_first", right_on="director_name", how="left")
df = df.drop(columns=["director_first", "director_name"])

director.to_csv("director.csv", index=False)
df.to_csv("metaclean_v2.csv", index=False)

print(director.shape)
print(df[["movie_id", "director_id"]].head(3))
print(df.director_id.isna().sum())   # moet 0 zijn
```

Here I created the DIRECTOR table and linked it to MOVIE. In the ERD there is a 1:N relation between DIRECTOR and MOVIE, so one director can have multiple movies, and that means the foreign key has to go on the movie side. So I don't need a separate junction table for this one, I just add a director_id column to the movie table, which is an FK that points to DIRECTOR. The movie_id stays the PK of MOVIE. First I checked how many movies have more than one director, and it turns out this is only the case for 2 out of the 11,332 movies. Because it's such a small amount, I decided to just take the first director that is mentioned for those movies, so we can stay within the 1:N structure of the ERD. This is a small deviation from the raw data, so I'm gonna mention it in the project document as well.

After that I took all the unique director names and gave each of them their own director_id (1, 2, 3 etc.), which gives me the DIRECTOR table with director_id and director_name. This resulted in 6,126 unique directors. Then I merged the director_id back into the main dataframe on the director name and dropped the helper columns afterwards, so the name only lives in the DIRECTOR table and not in MOVIE as well. I also checked that every one of the 11,332 movies got a director_id and that none of them is missing. In the end I saved two files: director.csv for the DIRECTOR table and metaclean_v2.csv, which is the movie data with the extra director_id column.

```python
df = pd.read_csv('metaclean_v2.csv')   

exploded = (
    df[["movie_id", "genre"]]
    .assign(genre=df["genre"].str.split(","))
    .explode("genre")
)
exploded["genre"] = exploded["genre"].str.strip()
exploded = exploded.dropna(subset=["genre"])
exploded = exploded[exploded["genre"] != ""]

print(len(exploded), "rows before removing duplicates")
print(exploded.duplicated().sum(), "duplicate rows")

dupes = exploded[exploded.duplicated(keep=False)].drop_duplicates()
print(dupes.merge(df[["movie_id", "title", "genre"]], on="movie_id",
                  suffixes=("_split", "_original")))

exploded = exploded.drop_duplicates()

print(len(exploded), "rows after removing duplicates")  

unique_genres = sorted(exploded["genre"].unique())
genre = pd.DataFrame({
    "genre_id": range(1, len(unique_genres) + 1),
    "genre": unique_genres
})

movie_genre = exploded.merge(genre, on="genre")[["movie_id", "genre_id"]]

genre.to_csv("genre.csv", index=False)
movie_genre.to_csv("movie_genre.csv", index=False)

print(genre.shape)
print(movie_genre.shape)
print(movie_genre.duplicated().sum())
```

Here I created the GENRE table and the MOVIE_GENRE junction table. In the ERD these are two separate tables, because a movie can have more than one genre and a genre belongs to a lot of movies. That is a many-to-many relation, so I can't just put a genre_id column in the movie table like I did for director. I first checked the genre column and there are no missing values. It turns out that 7,871 movies have more than one genre, and they are all in one text cell separated by commas. So I split the genre column on the comma and gave every genre of a movie its own row.

While doing this I noticed that for 4 movies (for example Anatomy of Hell and Intimacy) the genre Adult is listed twice in the same cell. If I leave this as it is, the junction table gets duplicate rows and that would give an error in PostgreSQL as soon as I set movie_id and genre_id together as the primary key. Therefore I removed these duplicates before building the tables.

After that I took all the unique genres and gave each of them their own genre_id (1, 2, 3 etc.), which gives me the GENRE table with genre_id and genre. This resulted in 27 unique genres. Then I merged this back with the split data on the genre name, so I could link every movie_id to the right genre_id. This is the MOVIE_GENRE junction table, and it has 27,029 rows. I also checked that there are no duplicate rows left in it and that all 11,332 movies are in there. In the end I saved two files: genre.csv for the GENRE table and movie_genre.csv for the junction table.

```python
df = pd.read_csv('metaclean_v2.csv')
sales = pd.read_csv('sales_cleaned.csv')

print(sales["match_key"].isna().sum(), "sales rows without match_key")
print(sales["match_key"].dropna().duplicated().sum(), "duplicate match_keys with a title")

movie_sales = (
    df[["movie_id", "match_key"]]
    .merge(sales[["sales_id", "match_key"]], on="match_key")
    [["movie_id", "sales_id"]]
)

print(movie_sales.shape)
print(movie_sales["movie_id"].is_unique)
print(movie_sales["sales_id"].is_unique)
print(movie_sales.duplicated().sum())
print((~sales["match_key"].isin(df["match_key"])).sum(), "sales rows without a match in meta")

movie_sales.to_csv("movie_sales.csv", index=False)
```

Here I created the MOVIE_SALES junction table that connects MOVIE to SALES. The two datasets don't share an ID, so I used the match_key (the cleaned title combined with the release year) that I created earlier in both files. Before joining I checked the sales data and found 8 rows without a title, and therefore without a match_key. These can't be matched to any movie, so they simply drop out of the join. I also checked that no match_key with a title appears more than once in the sales data, so a movie can't accidentally get linked to multiple sales rows.

After that I merged movie_id and match_key from the movie data with sales_id and match_key from the sales data, and kept only movie_id and sales_id. This resulted in 7,978 rows. I checked that there are no duplicate rows in it and that both movie_id and sales_id only appear once, so for now it works as a 1:1 link, but it stays a junction table because that is how it is drawn in the ERD. One thing to keep in mind is that only 7,978 of the 30,612 sales rows could be linked to a movie. The other 22,634 sales rows stay in the SALES table but have no row in MOVIE_SALES, which means the analysis can only use the movies that are linked. In the end I saved the result as movie_sales.csv.

```python
df = pd.read_csv('metaclean_v2.csv')
sales = pd.read_csv('sales_cleaned.csv')

runtime_sales = (
    df[["movie_id", "match_key"]]
    .merge(sales[["match_key", "runtime"]], on="match_key")
    [["movie_id", "runtime"]]
    .rename(columns={"runtime": "runtime_sales"})
)

movie =df.merge(runtime_sales, on="movie_id", how="left")
movie = movie.rename(columns={"runtime": "runtime_meta"})

movie = movie[["movie_id", "director_id", "url", "title", "studio", "rating", "runtime_meta", "runtime_sales", "metascore", "userscore", "RelDate"]]

movie["runtime_meta"] = movie["runtime_meta"].astype("Int64")
movie["runtime_sales"] = movie["runtime_sales"].round().astype("Int64")

movie.insert(2, "actor_id", pd.array([pd.NA] * len(movie), dtype="Int64"))

movie.to_csv("movie.csv", index=False)

print(movie.shape)
print(movie["movie_id"].is_unique)
print(movie["runtime_sales"].notna().sum(), "movies with a runtime from sales")
print(movie["actor_id"].isna().all())
print(movie.columns.tolist()) 
```

Here I created the MOVIE table. This is basically the metaclean data, but then reduced to only the columns that are in the ERD. The columns cast, director, genre, summary, awards, title_clean and match_key are dropped, because cast, director and genre now live in their own tables and the rest are only helper columns or text that we don't need in the database. I also renamed runtime to runtime_meta, because the ERD has two runtime columns, one from Metacritic and one from the sales data.

For runtime_sales I used the match_key again to link every movie to its row in the sales data and took the runtime from there. This resulted in 6,367 movies with a runtime_sales. The other movies get an empty value, because either the movie has no match in the sales data or the sales data has no runtime for it. I don't think this is a problem, but it is good to keep in mind if we use this column in the analysis. Both runtimes are also converted to whole numbers, because the ERD defines them as integers.
